```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
def plotRevenue(plotData,title='Revenue',savename=''):
  xAxis=plotData.index.to_numpy()
  yAxis=[plotData['RevenueLossbyUpgrade'].to_numpy()/1000,plotData['RevenueLossbyCancel'].to_numpy()/1000,plotData['PotentialRevenue'].to_numpy()/1000]
  plt.close()
  plt.figure(dpi=1200,figsize=(20,6))
  plt.stackplot(xAxis,yAxis, labels=['Revenue loss from free upgrades ','Revenue loss from last-minute cancellations ','Potential Revenue'], alpha=0.4 )
  plt.legend(loc="upper left", fontsize=10 , ncol = 3)
  plt.title(title)
  plt.yticks(np.arange(100, np.max(yAxis), 100), fontsize=10)
  plt.xlabel('Number of the week')
  plt.ylabel('Potential Revenue (Thousand)')
  plt.xticks(xAxis[::1], fontsize=10, horizontalalignment='center')
  plt.xlim(xAxis[0], xAxis[-1])
  plt.gca().spines["top"].set_alpha(0)
  plt.gca().spines["bottom"].set_alpha(.3)
  plt.gca().spines["right"].set_alpha(0)
  plt.gca().spines["left"].set_alpha(.3)
  if savename!='':
    plt.savefig(savename+".png",transparent=True)
  plt.show()
```

```python
def plotChange(before,after,label,legends,title='Revenue Loss Comparison',savename=''):
  xAxis=before.index.to_numpy()
  yAxis=[after[label].to_numpy(dtype=float),before[label].to_numpy(dtype=float)-after[label].to_numpy(dtype=float)]
  #yAxis=[np.arange(1,54,1),np.arange(1,54,1)]
  plt.close()
  plt.figure(dpi=1200,figsize=(20,6))
  plt.stackplot(xAxis,yAxis, labels=legends, alpha=0.4 )
  plt.legend(loc="upper left", fontsize=10 , ncol = 2)
  plt.title(title)
  #plt.yticks(np.arange(100, np.max(yAxis), 100), fontsize=10)
  plt.xlabel('Number of the week')
  plt.ylabel('Revenue Loss')
  plt.xticks(xAxis[::1], fontsize=10, horizontalalignment='center')
  plt.xlim(xAxis[0], xAxis[-1])
  plt.gca().spines["top"].set_alpha(0)
  plt.gca().spines["bottom"].set_alpha(.3)
  plt.gca().spines["right"].set_alpha(0)
  plt.gca().spines["left"].set_alpha(.3)
  if savename!='':
    plt.savefig(savename+".png",transparent=True)
  plt.show()
```

```python
def plotPotentialChange(before,after,label,legends,title='Revenue Loss Comparison',savename=''):
  xAxis=before.index.to_numpy()
  yAxis=[after[label].to_numpy(dtype=float),before[label].to_numpy(dtype=float)-after[label].to_numpy(dtype=float)]
  plt.close()
  plt.figure(dpi=1200,figsize=(20,6))
  plt.stackplot(xAxis,yAxis, labels=legends, alpha=0.4 )
  plt.legend(loc="upper left", fontsize=10 , ncol = 2)
  plt.title(title)
  plt.xlabel('Number of the week')
  plt.ylabel('Potential Revenue')
  plt.ticklabel_format(style='plain')
  plt.xticks(xAxis[::1], fontsize=10, horizontalalignment='center')
  #plt.yticks(np.arange(np.min(yAxis), np.max(yAxis), 100), fontsize=10)
  plt.xlim(xAxis[0], xAxis[-1])
  plt.gca().spines["top"].set_alpha(0)
  plt.gca().spines["bottom"].set_alpha(.3)
  plt.gca().spines["right"].set_alpha(0)
  plt.gca().spines["left"].set_alpha(.3)
  if savename!='':
    plt.savefig(savename+".png",transparent=True)
  plt.show()
```

```python
rawData=pd.read_excel('R2_final_hotels_Clean.xlsx')
```

```python
plotData=rawData.groupby('ArrivalWeekNumber')['RevenuebyStays','RevenueLossbyUpgrade','RevenueLossbyCancel','PotentialRevenue'].sum()
```

```python
plotRevenue(plotData,savename='Revenue Decomposition')
```

```python
upgradeData=pd.read_excel('R2_final_hotels_upgradepremium.xlsx')
```

```python
upgradeData['RevenueLossbyUpgrade']=upgradeData['RevenueLossbyUpgrade']-upgradeData['UpgradePremium']*(upgradeData['StaysInWeekendNights']+upgradeData['StaysInWeekNights'])
```

```python
upgradeData['RevenueLossbyUpgrade']=upgradeData['RevenueLossbyUpgrade'].clip(lower=0)
plotupgradeData=upgradeData.groupby('ArrivalWeekNumber')['RevenuebyStays','RevenueLossbyUpgrade','RevenueLossbyCancel','PotentialRevenue'].sum()
```

```python
cancelData=pd.read_excel('R2_final_hotels_cancelpremium.xlsx')
```

```python
cancelData['RevenueLossbyCancel']=cancelData['RevenueLossbyCancel']-cancelData['CancelPremium']*(cancelData['StaysInWeekendNights']+cancelData['StaysInWeekNights'])
```

```python
cancelData['RevenueLossbyCancel']=cancelData['RevenueLossbyCancel'].clip(lower=0)
plotcancelData=cancelData.groupby('ArrivalWeekNumber')['RevenuebyStays','RevenueLossbyUpgrade','RevenueLossbyCancel','PotentialRevenue'].sum()
```

```python
print("orginal Revenue Loss by Upgrade: "+str(sum(plotData['RevenueLossbyUpgrade'])))
print("after Revenue Loss by Upgrade: "+str(sum(plotupgradeData['RevenueLossbyUpgrade'])))
```

```python
print("orginal Revenue Loss by Cancel: "+str(sum(plotData['RevenueLossbyCancel'])))
print("after Revenue Loss by Cancel: "+str(sum(plotcancelData['RevenueLossbyCancel'])))
```

```python
plotChange(plotData,plotupgradeData,'RevenueLossbyUpgrade',['Revenue Loss by Upgrade with Upgrade Premium','Reduction of Revenue Loss by Upgrade'],title='Revenue Loss by Upgrade',savename='Revenue Loss by Upgrade')
```

```python
plotChange(plotData,plotcancelData,'RevenueLossbyCancel',['Revenue Loss by Cancel with Cancel Premium','Reduction of Revenue Loss by Cancel'],title='Revenue Loss by Cancel',savename='Revenue Loss by Cancel')
```

```python
newData=rawData.copy(deep=True)
newData1=rawData.copy(deep=True)
newData2=rawData.copy(deep=True)
newData['PotentialRevenue']=rawData['PotentialRevenue']+(rawData['RevenueLossbyUpgrade']-upgradeData['RevenueLossbyUpgrade'])+(rawData['RevenueLossbyCancel']-cancelData['RevenueLossbyCancel'])
newData1['PotentialRevenue']=rawData['PotentialRevenue']+(rawData['RevenueLossbyUpgrade']-upgradeData['RevenueLossbyUpgrade'])
newData2['PotentialRevenue']=rawData['PotentialRevenue']+(rawData['RevenueLossbyCancel']-cancelData['RevenueLossbyCancel'])
plotnewData=newData.groupby('ArrivalWeekNumber')['RevenuebyStays','RevenueLossbyUpgrade','RevenueLossbyCancel','PotentialRevenue'].sum()
plotnewData1=newData1.groupby('ArrivalWeekNumber')['RevenuebyStays','RevenueLossbyUpgrade','RevenueLossbyCancel','PotentialRevenue'].sum()
plotnewData2=newData2.groupby('ArrivalWeekNumber')['RevenuebyStays','RevenueLossbyUpgrade','RevenueLossbyCancel','PotentialRevenue'].sum()
```

```python
print("orginal Potential Revenue: "+str(sum(plotData['PotentialRevenue'])))
print("after Potential Revenue with Upgrade&Cancel Premium: "+str(sum(plotnewData['PotentialRevenue'])))
print("after Potential Revenue with Upgrade Premium: "+str(sum(plotnewData1['PotentialRevenue'])))
print("after Potential Revenue with Cancel Premium: "+str(sum(plotnewData2['PotentialRevenue'])))
```

```python
plotPotentialChange(plotnewData,plotData,'PotentialRevenue',['Potential Revenue with Upgrade&Cancel Premium','Increase of Potential Revenue'],title='Potential Revenue',savename='Potential Revenue')
```

```python
plotPotentialChange(plotnewData1,plotData,'PotentialRevenue',['Potential Revenue with Upgrade Premium','Increase of Potential Revenue'],title='Potential Revenue',savename='Potential Revenue with Upgrade Premium')
```

```python
plotPotentialChange(plotnewData2,plotData,'PotentialRevenue',['Potential Revenue with Cancel Premium','Increase of Potential Revenue'],title='Potential Revenue',savename='Potential Revenue with Cancel Premium')
```
